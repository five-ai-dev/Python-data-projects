```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt

#显示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

#创建子图
# figure: 画布对象; axes；子图数组（里面存放的是Axes类型的对象）
figure, axes =plt.subplots(nrows=1, ncols=2, figsize=(20,6), dpi=100) # nrows: 行数; ncols: 列数, figsize(): 画布大小,宽高, dpi: 分辨率

# 图一；柱状图(世界石油储备)
countries = ['美国', '中国', '英国', '日本', '德国']
values = [97.0, 80.0, 43.0, 40.0, 29.0]

axes1: Axes = axes[0] # 第一个子图
axes1.bar(countries, values, width=0.4, color='r')# 柱状图
axes1.set_title('世界石油储备', fontsize=17) # 设置子图标题
axes1.set_xlabel('国家', fontsize=15)# x轴标签
axes1.set_ylabel('石油储备(亿吨)', fontsize=15)
axes1.grid(linestyle='--', alpha=0.3)# 网格


# 图二: 饼状图(世界人口占比)
countries = ['中国', '印度', '美国', '印尼', '巴基斯坦', '其他']
values = [21.2, 23.1, 5.6, 3.7, 2.93, 43.47 ]
axes2: Axes = axes[1] #第二个子图
axes2.pie(values, labels=countries, autopct='%1.1f%%',) # 饼状图
axes2.set_title('世界人口占比', fontsize=17) #标题
axes2.legend(loc = 'lower center',ncols = 5, bbox_to_anchor=(0.5, -0.1))# 显示图例, ncols: 列数, bbox_to_anchor: 图例位置, (0.5 控制水平方位, -0.1 控制垂直方位)

#保存图片
figure.savefig('data/01.png')



plt.show()# 显示图表

```
