```python
import matplotlib.pyplot as plt
import random

from numpy.ma.core import arange

#解决中文显示问题
plt.rcParams['font.sans-serif'] = ['SimHei']#设置中文字体为黑体

# 目标:绘制曲线图
x = [i for i in range(1,25)]
y_nj = [random.randint(5,15) for i in x]
y_gs = [random.randint(10,20) for i in x]

plt.figure(figsize=(10,5))# 设置画布大小，宽为10，高为5
plt.plot(x, y_nj, label='南京')# 折线图-->如果没有画布， 会自动创建一个画布
plt.plot(x, y_gs, label='甘肃')

# 设置折线图的详细信息
plt.title("气温变化折线图", fontsize = 15)# 设置标题
plt.xlabel("时间") # x轴标签
plt.ylabel("温度") # y轴标签

# plt.xticks(x[::2])
plt.xticks(x) # 设置x轴刻度
y_ticks = [i for i in range(5,21)]
plt.yticks(y_ticks) # 设置y轴刻度
plt.grid(linestyle = '--', alpha = 0.4) # 显示网格
plt.legend(loc = 'upper right') # 显示图例

plt.show() # 显示图表
```
