```python
import pandas as pd

pd.set_option('display.min_rows', 30)# 显示行数
# 1.读取数据
df = pd.read_csv('data/sales.csv')

# 2. 数据清洗
# 2.1 缺失值处理
df.isnull()# 查看缺失值

# 2.1.1 删除缺失值
# df.dropna()# 删除缺失值所在行axis=0
# df = df.dropna(axis=1) # 删除缺失值所在列axis=1
# df
# 2.1.2 填充缺失值
df.fillna('--')# 填充缺失值
df.ffill()# 填充缺失值,用前一个值填充
df.bfill()# 填充缺失值,用后一个值填充


```

```python
#2.2重复值处理
#2.2.1 查看重复值
df.duplicated()# 查看重复值
df.duplicated(subset=['订单号'])# 查看重复值(根据订单号)

# 2.2.2 删除重复值
df.drop_duplicates(subset=['订单号'], keep=False)# keep = 'first'保留第一个, keep = 'last'保留最后一个, keep = False不保留(所有)


```

```python
# 2.3异常值处理
#2.3.1 查看异常值
df[df['单价'] < 0]

# 2.3.2 删除异常值
df.drop(df[df['单价']<0].index)

# 2.3.3 修复异常值
df['单价'] = df['单价'].abs()# 绝对值
df

# 2.4 数据格式处理
df["订单日期"] = df["订单日期"].str.replace("/", "-")
df
```
