```python
df3 = pd.read_csv('data/sales.csv')
#配置项，显示所有数据
# pd.set_option('display.max_rows', None)


#数据过滤 --> df[filter]
#1. 获取销售数量 >= 10
df3[df3['销售数量']>=10]

# 2. 获取产品类别为食品或图书的订单数据
df3[df3['产品类别'].isin(['食品', '图书'])]

# 3. 获取 单价在100-200之间的订单数据--->范围:between-->默认包含边界（inclusive = 'left)只包含左
df3[df3['单价'].between(100, 200)]

# 4. 获取销售数量>=8， 并且 单价>=100的订单数据---->多条件(并且:&, 或者:|)
df3[(df3['销售数量']>=8) & (df3['单价']>=100)]

# 5. 获取 产品类别为服装/食品， 支付方式为支付宝/微信支付的订单数据
df3[(df3['产品类别'].isin(['服装', '食品'])) & (df3['支付方式'].isin(['支付宝', '微信支付']))]
```
